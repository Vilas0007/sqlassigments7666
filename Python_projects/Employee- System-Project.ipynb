```python
class Employee:
     def __init__(self,name,age,salary):
          self.name =name
          self.age = age
          self.salary = salary
     
     def emp_info_fun(self):
          print(f"name : {self.name} \nage : {self.age} \nsalary : {self.salary}")
          print("\n======================")
               

class EmployeesManager():
     def __init__(self):
          self.store_list = []
          
          
     def add_new_emp_fun(self,emp):
          self.store_list.append(emp)
     
     def delete_emp_fun(self,start,end):
          for i in self.store_list[:]:      
               if start <= i.age <= end:
                    self.store_list.remove(i)
           
               
     def find_emp_fun(self,name):
          for i in self.store_list:
               if i.name == name:
                    print(f"name {i.name}")         
          return None 
     def update_fun(self,name,new_salary):
          for i in self.store_list:
               if i.name == name:
                    i.salary =new_salary
                    break
    
     def show_all_emp_fun(self):
          for emp in self.store_list:
               emp.emp_info_fun()
               
class FrontendManager:
     def __init__(self):
          self.manager = EmployeesManager()

     def start_fun(self):
          while True:
               print("\n1. Add Employee")
               print("2. Show All Employees")
               print("3. Find Employee")
               print("4. Update Salary")
               print("5. Delete Employees by Age")
               print("6. Exit")

               choice = input("Enter choice: ")

               if choice == "1":
                    name = input("Name: ")
                    age = int(input("Age: "))
                    salary = int(input("Salary: "))
                    emp = Employee(name, age, salary)
                    self.manager.add_new_emp_fun(emp)
                    print("\n======================")
               elif choice == "2":
                    self.manager.show_all_emp_fun()
                    print("\n======================")

               elif choice == "3":
                    name = input("Enter name: ")
                    self.manager.find_emp_fun(name)

               elif choice == "4":
                    name = input("Name: ")
                    salary = int(input("New Salary: "))
                    self.manager.update_fun(name, salary)
                    print("\n======================")
                    
               elif choice == "5":
                    start = int(input("Start Age: "))
                    end = int(input("End Age: "))
                    self.manager.delete_emp_fun(start, end)
                    
               elif choice == "6":
                    print("\n======================")
                    break
app = FrontendManager()
app.start_fun()
```
