```python
#5. Write a Python program to create a class representing a binary search tree.
#Include methods for inserting and searching for elements in the binary tree.
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


class BST:
    def __init__(self):
        self.root = None

    def insert(self, data):
        def insert_node(root, data):
            if root is None:
                return Node(data)
            if data < root.data:
                root.left = insert_node(root.left, data)
            else:
                root.right = insert_node(root.right, data)
            return root

        self.root = insert_node(self.root, data)

    def search(self, key):
        def search_node(root, key):
            if root is None:
                return False
            if root.data == key:
                return True
            elif key < root.data:
                return search_node(root.left, key)
            else:
                return search_node(root.right, key)

        result = search_node(self.root, key)
        print(f"Element {key} found: {result}")

bst = BST()
bst.insert(10)
bst.insert(5)
bst.insert(15)

bst.search(5)
bst.search(20)
```

```python
#6. Write a Python program to create a class representing a stack data structure.
#Include methods for pushing and popping elements.
class Stack:
    def __init__(self):
        self.stack = []

    def push(self, value):
        self.stack.append(value)
        print(f"Pushed: {value}")

    def pop(self):
        if not self.stack:
            print("Stack is empty")
        else:
            removed = self.stack.pop()
            print(f"Popped: {removed}")


s = Stack()
s.push(10)
s.push(20)
s.pop()
```

```python
#7. Write a Python program to create a class representing a linked list data structure.
#Include methods for displaying linked list data, inserting and deleting nodes.
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
        else:
            temp = self.head
            while temp.next:
                temp = temp.next
            temp.next = new_node
        print(f"Inserted: {data}")

    def delete(self, key):
        temp = self.head

        if temp and temp.data == key:
            self.head = temp.next
            print(f"Deleted: {key}")
            return

        prev = None
        while temp and temp.data != key:
            prev = temp
            temp = temp.next

        if temp is None:
            print(f"Element {key} not found")
        else:
            prev.next = temp.next
            print(f"Deleted: {key}")

    def display(self):
        temp = self.head
        print("Linked List:", end=" ")
        while temp:
            print(f"{temp.data} ->", end=" ")
            temp = temp.next
        print("None")


ll = LinkedList()
ll.insert(10)
ll.insert(20)
ll.insert(30)

ll.display()

ll.delete(20)
ll.display()
```

```python
'''
14. Write a program that uses datetime module within a class.
Enter manufacturing date and expiry date of the product.
The program must display the years, months and days that are left for expiry.
'''

from datetime import datetime

class Product:
    def __init__(self, mfg_date, exp_date):
        # Convert string to datetime object
        self.mfg_date = datetime.strptime(mfg_date, "%Y-%m-%d")
        self.exp_date = datetime.strptime(exp_date, "%Y-%m-%d")

    def time_left(self):
        today = datetime.today()
        
        if today > self.exp_date:
            print("Product already expired ❌")
            return
        
        diff = self.exp_date - today
        
        # Convert days into years, months, days (approx)
        days = diff.days
        years = days // 365
        months = (days % 365) // 30
        remaining_days = (days % 365) % 30
        
        print(f"Time left for expiry: {years} years, {months} months, {remaining_days} days")


mfg = input("Enter Manufacturing Date (YYYY-MM-DD): ")
exp = input("Enter Expiry Date (YYYY-MM-DD): ")

p = Product(mfg, exp)
p.time_left()


```

```python
'''
19. Write a Python class Restaurant with attributes like menu_items,
 book_table, and customer_orders, and methods like add_item_to_menu,
 book_tables, and customer_order.

	Perform the following tasks now:

	Now add items to the menu.
	Make table reservations.
	Take customer orders.
	Print the menu.
	Print table reservations.
	Print customer orders.
	Note: Use dictionaries and lists to store the data.
'''
class Restaurant:
    def __init__(self):
        self.menu_items = {}       
        self.book_table = []         
        self.customer_orders = []     

    def add_item_to_menu_fun(self, item, price):
        self.menu_items[item] = price

    def book_tables_fun(self, table_number):
        self.book_table.append(table_number)

    def customer_order_fun(self, order):
        self.customer_orders.append(order)

    def print_menu_fun(self):
        print("\nMenu:")
        for item, price in self.menu_items.items():
            print(item, ":", price)

    def print_tables_fun(self):
        print("\nBooked Tables:", self.book_table)

    def print_orders_fun(self):
        print("\nCustomer Orders:", self.customer_orders)



r1 = Restaurant()
r1.add_item_to_menu_fun("Pizza", 200)
r1.add_item_to_menu_fun("Burger", 100)

r1.book_tables_fun(1)
r1.book_tables_fun(2)

r1.customer_order_fun("Pizza")
r1.customer_order_fun("Burger")

r1.print_menu_fun()
r1.print_tables_fun()
r1.print_orders_fun()
```

```python
#21. Write a Python class Inventory with attributes like item_id,
#item_name, stock_count, and price, and methods like add_item, update_item,
# and check_item_details.
class Inventory:
    def __init__(self):
        self.items = {}   

    def add_item_fun(self, item_id, item_name, stock_count, price):
        self.items[item_id] = {
            "name": item_name,
            "stock": stock_count,
            "price": price
        }
        print(f"{item_name} added successfully")

    def update_item_fun(self, item_id, new_name=None, new_stock=None, new_price=None):
        if item_id in self.items:
            if new_name:
                self.items[item_id]["name"] = new_name
            if new_stock is not None:
                self.items[item_id]["stock"] = new_stock
            if new_price is not None:
                self.items[item_id]["price"] = new_price
            print("Item updated successfully")
        else:
            print("Item not found")

    def check_item_details_fun(self, item_id):
        if item_id in self.items:
            print("Item Details:", self.items[item_id])
        else:
            print("Item not found")

i1 = Inventory()

i1.add_item_fun(1, "Laptop", 10, 40000)
i1.add_item_fun(2, "Mouse", 50, 500)

i1.update_item_fun(1, new_price=38000)

i1.check_item_details_fun(1)

     
```
