7. Sort a list of alphanumeric strings based on product value of numeric character in it. If in any string there is no numeric character take it's product value as 1.
	Input:
	['1ac21', '23fg', '456', '098d','1','kls']
	Output:
	['456', '23fg', '1ac21', '1', 'kls', '098d']

	

```python
def product_of_digits(s):
    prod = 1
    found_digit = False
    
    for ch in s:
        if ch.isdigit():
            prod *= int(ch)
            found_digit = True
    
    return prod if found_digit else 1


def sort_strings(lst):
    result = sorted(lst, key=product_of_digits)
    print(f"Sorted List: {result}")


# Example
data = ['1ac21', '23fg', '456', '098d','1','kls']
sort_strings(data)
```

```python
# 11. Write a program to count unique number of vowels using sets in a given string. Lowercase and 
# upercase vowels will be taken as different.
# 	Input:
# 	Str1 = "hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX"
# 	Output:
# 	No of unique vowels-6

def count_unique_vowels(s):
    vowels = set("aeiouAEIOU")   # both lowercase & uppercase
    found_vowels = set()

    for ch in s:
        if ch in vowels:
            found_vowels.add(ch)

    print(f"Unique vowels: {found_vowels}")
    print(f"No of unique vowels: {len(found_vowels)}")


Str1 = "hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX"
count_unique_vowels(Str1)








```

```python
# 12. Intersection of two lists. Intersection of two list means we need to take all those elements which are common to both of the initial lists and store them into another list. Only use using list-comprehension.
# 	Example 1:

# 	Input:

# 	lst1 = {15, 9, 10, 56, 23, 78, 5, 4, 9}
# 	lst2 = {9, 4, 5, 36, 47, 26, 10, 45, 87}
# 	Output:

# 	[9, 10, 4, 5]
# 	Example 2:

# 	Input:

# 	lst1 = {4, 9, 1, 17, 11, 26, 28, 54, 69}
# 	lst2 = {9, 9, 74, 21, 45, 11, 63, 28, 26}
# 	Output:

# 	[9, 11, 26, 28]
def intersection(lst1, lst2):
    return [x for x in lst1 if x in lst2]


# Example 1
lst1 = [15, 9, 10, 56, 23, 78, 5, 4, 9]
lst2 = [9, 4, 5, 36, 47, 26, 10, 45, 87]

print(f"Intersection: {intersection(lst1, lst2)}")


# Example 2
lst1 = [4, 9, 1, 17, 11, 26, 28, 54, 69]
lst2 = [9, 9, 74, 21, 45, 11, 63, 28, 26]

print(f"Intersection: {intersection(lst1, lst2)}")

Intersection: [9, 10, 5, 4, 9]
Intersection: [9, 11, 26, 28]
```

```python
# 15.  Write a Python program to print the even numbers from a given list.
# 		Sample List : [1, 2, 3, 4, 5, 6, 7, 8, 9]
# 		Expected Result : [2, 4, 6, 8]
		
def even_numbers(lst):
    result = []
    
    for num in lst:
        if num % 2 == 0:
            result.append(num)
    
    print(f"Even numbers: {result}")


data = [1, 2, 3, 4, 5, 6, 7, 8, 9]
even_numbers(data)

```

```python
# 16. Write a Python function to check whether a number is perfect or not.¶
# 	A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

# 	Example :

# 	The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
# 	Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.

# 	The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.

def is_perfect_fun(num):
    if num <= 0:
        print(f"{num} is not a Perfect Number")
        return

    sum_div = 0

    # Find proper divisors
    for i in range(1, num):
        if num % i == 0:
            sum_div += i

    if sum_div == num:
        print(f"{num} is a Perfect Number ")
    else:
        print(f"{num} is NOT a Perfect Number ")


# Example
is_perfect_fun(6)
is_perfect_fun(28)
is_perfect_fun(10)
```

```python
# 17. Write a Python function to concatenate any no of dictionaries to create a new one.
# 	Sample Dictionary :
# 	dic1={1:10, 2:20}
# 	dic2={3:30, 4:40}
# 	dic3={5:50,6:60}
# 	Expected Result : {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}

def merge_dicts_fun(*dicts):
    result = {}
    
    for d in dicts:
        result.update(d)
    
    return result


# Example
dic1 = {1: 10, 2: 20}
dic2 = {3: 30, 4: 40}
dic3 = {5: 50, 6: 60}

final_dict = merge_dicts_fun(dic1, dic2, dic3)

print(f"Merged Dictionary: {final_dict}")
     
```

```python
# 18. Write a python function that receives a list of integers and prints out a histogram of bin size 10

# 	Input:
# 	[13,42,15,37,22,39,41,50]
# 	Output:
# 	{11-20:2,21-30:1,31-40:2,41-50:3}
	
def histogram(nums):
    bins = {}

    for num in nums:
        # Find range start and end
        start = (num // 10) * 10 + 1
        end = start + 9

        key = f"{start}-{end}"

        # Count frequency
        bins[key] = bins.get(key, 0) + 1

    print(bins)


# Example
data = [13, 42, 15, 37, 22, 39, 41, 50]
histogram(data)
```
