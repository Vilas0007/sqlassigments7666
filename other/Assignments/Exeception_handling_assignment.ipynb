```python
#1. Write a Python program to handle a ZeroDivisionError exception when dividing a number by zero.
num = int(input("enter divisor : "))
num_list = [1,24,5,9,3]
for i in num_list:
    try:
        print(i/num)
    except ZeroDivisionError as z:
        print(f"{z} : try again with another number except 0")
```

```python
#2. Write a Python program that prompts the user to input an integer and raises
#a ValueError exception if the input is not a valid integer
try:
    num =int(input("enter any number : "))
    if num % 2==0:
      print(f"{num} is a even number")
    else:
      print(f"{num} is odd number")
except ValueError as v :
  print(f"{v} : please input any valid integer number")
```

```python
#3. Write a Python program that opens a file and handles a FileNotFoundError
# an exception if the file does not exist.
try:
  file = open("python_test1.tst","r")
  print(file.read())
  file.close()
except FileNotFoundError as f:
  print(f"{f} : file not found")
```

```python
#6. Write a Python program that executes an operation on a list and handles an IndexError exception if the index is out of range.
try :
  num_list = [1,2,3,4,5]
  for i in range(len(num_list)+1):
    print(num_list[i])
except IndexError as i:
  print(i)
```
