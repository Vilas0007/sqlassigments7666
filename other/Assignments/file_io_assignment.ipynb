```python
#7. Write a python program to find the longest words.
with open("sample2.txt","r") as f:
     data = f.read().split()
     # print(data)
     longest_word =""
     for i in data:
          if len(i) > len(longest_word):
               longest_word = i
          
print("Longest word:", longest_word)
```

```python
#9. Write a Python program to count the frequency of words in a file.
def frequency_of_word_fun(word):
     with open ("sample2.txt","r") as f:
          data = f.read().split()
          # word = "hi"
          count = 0
          for i in data:
               if i.lower() == word.lower():
                    count += 1
          print(f"frequency of {word} is {count}")
frequency_of_word_fun("ANIruddHa")
```
